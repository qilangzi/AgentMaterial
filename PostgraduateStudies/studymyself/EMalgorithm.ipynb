- 高斯混合模型
    - 对于一组数据可列似然函数：
        - $l_{\pi, \mu, \Sigma}(x) = \sum_{i=1}^{N} \log \left( \sum_{k=1}^{K} \pi_k N(x_i \mid \mu_k, \Sigma_k) \right)$
    - 高斯分布的概率密度函数 
        - $f(x) = \frac{1}{\sqrt{2\pi}\sigma} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$

        - $L(x) = \prod_{i=1}^{n} \frac{1}{\sqrt{2\pi}\sigma} e^{-\frac{(x_i-\mu)^2}{2\sigma^2}}$
## EM算法

- 通过不断的下界函数求紫色函数最大值 
    - 尽可能紧的下界
        - $l(\theta) = \sum_{i=1}^{m} \log \sum_{z} p(x, z; \theta) = \sum_{i=1}^{m} \log \sum_{z^{(i)}} p(x^{(i)}, z^{(i)}; \theta)= \sum_{i=1}^{m} \log \sum_{z^{(i)}} Q_i(z^{(i)}) \frac{p(x^{(i)}, z^{(i)}; \theta)}{Q_i(z^{(i)})}\geq \sum_{i=1}^{m} \sum_{z^{(i)}} Q_i(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)}; \theta)}{Q_i(z^{(i)})}$
    - 该函数成正比$\frac{p(x^{(i)}, z^{(i)}; \theta)}{Q_i(z^{(i)})} = c$
